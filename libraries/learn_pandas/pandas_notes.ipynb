## Импортирование Pandas

```python
import pandas as pd
```

## Основные структуры данных

**DataFrame** - таблица (как в Excel)

```python
data_frame = pd.DataFrame({
    'имя': ['Анна', 'Борис', 'Вася'],
    'возраст': [25, 30, 35],
    'город': ['Москва', 'СПб', 'Казань']
})

print(data_frame)
```

```python
series = pd.Series(data=[1, 3, 3, 4, 5], index=["a", "b", "c", "d", "e"])

print(series)
```

### Работа со столбцами

**Получение столбца**

`df["название столбца"]`

```python
body_mass = df_csv["body_mass_g"]

print(body_mass)
```

**Получение нескольких столбцов**

`df[["столбец_1", "столбец_2"]]`

```python
print(df_csv[["sex", "year"]])
```

**Список всех стобцов** - `df.columns`

```python
all_columns = df_csv.columns

print(all_columns)
```

### Фильтрация

**Условия**

Строки где вес больше 6000

`df[df["столбец"] условие (>, <, >=, <=)]`

```python
body_mass_more_6000 = df_csv[df_csv["body_mass_g"] >= 6000]

print(body_mass_more_6000)
```

**Несколько условий**

Вес больше 5800 и 2008 год

`df[df[(условие_1) & или | (условие_2)]]`

```python
mass_more_5800_and_2008 = df_csv[(df_csv["body_mass_g"] >= 5800) & (df_csv["year"] == 2007)]

print(mass_more_5800_and_2008)
```

## Часть используемые

### Создание DataFrame

**Из словаря** - `pd.DataFrame(словарь)`

```python
df_from_dict = pd.DataFrame(
    {
        "name": ["Paul", "Nikita"],
        "age": [22, 21]
    }
)

print(df_from_dict)
```

**Из списка списков** - `pd.DataFrame(список из списков, columns=["название_1", "название_2"])`

```python
df_from_list = pd.DataFrame(
    [["Paul", 20], ["Nikita", 23], ["Paul", 20]],
    columns=["name", "age"]
)

print(df_from_list)
```

### Работа с пропусками

**Проверка, где существуют пропуски** - `df.isna()`

True - нет данных

```python
data_gaps = df_csv.isna()

print(data_gaps)
```

**Удаление строк с пропусками** - `df.dropna()`

Возвращает frame без пропусков, но в исходном файле они *остаются*

```python
df_without_na = df_csv.dropna()

print(df_without_na.isna())
print(df_csv.isna())
```

**Заполнение пропусков** - `df.fillna(значение, которым заполняем пропуски)`

```python
df_with_filled_na = df_csv.fillna(100)

print(df_with_filled_na.head())
```

### Группировка и агрегация

**Группировка** - `df.groupby("по какой колонке группируем")["к какой колонке применяем функцию"].mean()`

Средний вес у девочек
True - условие верно (девочки)
False - условие верно (мальчики)

```python
mean_mass_from_female = df_without_na.groupby(df_without_na["sex"] == "female")["body_mass_g"].mean()

print(mean_mass_from_female)
```

**Несколько операций** - `df.groupby("по какой колонке группируем").agg({столбец_1: операция_1, столбец_2: операция_2})`

Средний вес у девочек и мальчиков + количество особей

```python
mean_mass_count_species = df_without_na.groupby("sex").agg({"body_mass_g": "mean", "species": "count"})

print(mean_mass_count_species)
```

### Сортировка

**Сортировка** - `df.sort_values("столбец", ascending=False(убывание))`

```python
sorted_from_year_df = df_csv.sort_values("year", ascending=False)  # по убыванию
sorted_from_mass_df = df_csv.sort_values("body_mass_g")

print(sorted_from_year_df)
print(sorted_from_mass_df)
```

### Добавление и удаление

**Добавление столбца** - `df["новый столбец"] = [значения, ...]`

```python
df_from_list["sallary"] = [10000, 50000]

print(df_from_list)
```

**Удаление столбца** - `df.drop(columns="столбец, который удаляем")`

```python
df_from_list = df_from_list.drop(columns="sallary")

print(df_from_list)
```

**Удаление строки** - `df.drop([строки], axis=0)`

```python
df_from_list = df_from_list.drop([1], axis=0)

print(df_from_list)
```

## Полезные действия

**Частота значений** - `df.value_counts()`

Сколько раз встречается каждая комбинация значений во всех столбцах

```python
freq_val = df_csv.value_counts()

print(freq_val)
```

**Уникальные значения** - `df.unique()`

Для Series

```python
unique_el = series.unique()

print(series)
print(unique_el)
```

**Дубликаты** - `df.dublicated()`

```python
dublicates = df_from_list.duplicated()

print(df_from_list)
print(dublicates)
```

**Удалить дубликаты** - `df.drop_dulicates()`

```python
print(df_from_list)

df_without_dublicate = df_from_list.drop_duplicates()

print(df_without_dublicate)
```

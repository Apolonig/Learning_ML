## Импортирование

```python
import matplotlib.pyplot as plt
```

## Базовый график

```python
coordinates_x = [1, 2, 4, 5, 6, 9, 10, 12, 13, 15, 16, 17, 22, 25]
coordinates_y = [2, 5, 7, 8, 9, 12, 14, 15, 16, 18, 20, 23, 25, 27]

plt.plot(coordinates_x, coordinates_y)
plt.show()
```

## Основный функции

### Линейный график

Строит линейный график по точкам x и y

```python
plt.plot(coordinates_x, coordinates_y)
plt.show()
```

### Точечный график - диаграмма  рассеивания

```python
plt.scatter(coordinates_x, coordinates_y)
plt.show()
```

### Столбчатая диаграмма


```python
plt.bar(coordinates_x, coordinates_y)
plt.show()
```

### Гистограмма

Сколько значений попадает в каждый интервал

```python
import numpy as np

data = np.array([10, 12, 15, 15, 18, 20, 22, 22, 25, 30])

plt.hist(data, bins=5)
plt.show()
```

### Круговая диаграмма - `plt.pie(значение категории, название категорий)` 

```python
value_categorise = np.array([30, 20, 10, 30, 5, 5])
lables_categorise = np.array(["Python", "Java", "C++", "JavaScript", "Go", "Rust"])

plt.pie(value_categorise, labels=lables_categorise)
plt.show()
```

## Оформление графиков

```python
plt.plot(coordinates_x, coordinates_y, label="Chart")

# Заголовок графика
plt.title("Line plot")

# Подписи осей
plt.xlabel("X")
plt.ylabel("Y")

# Легенда
# для этого надо указать lable при создании графика
plt.legend()

# Сетка на график
plt.grid()

# Услинение предела осей
plt.xlim(0, 30)
plt.ylim(0, 30)

plt.show()
```

## То что используется крайне часто

### Фигура с заданным размером

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(10, 5))

plt.plot(x, y)

plt.title("Линейный график")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

### Несколько графиков в одном изображении

```python
plt.subplot(2, 2, 1)
plt.plot(x, y)

plt.subplot(2, 2, 2)
plt.bar(x, y)

plt.show()
```

### Сохранение графика в файл

Поддержка - png, pdf, svg

```python
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(x, y)

plt.subplot(2, 2, 2)
plt.bar(x, y)
plt.savefig("chart.png")
```

## Стилизация линий

- color - цвет линии
- linestyle - стиль линии -, --, :, -.
- linewidth - толщина линии
- marker - маркер точек o, ^, *, s

```python
plt.plot(x, y, color="green", linestyle="--", linewidth=2, marker="o")

plt.show()
```

## Пример полноценного графика

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 15, 18]
y2 = [1, 2, 3, 4, 5]

plt.figure(figsize=(10, 8))

plt.plot(x, y1, label="Красная", color="red", marker="o", linestyle=":")
plt.plot(x, y2, label="Фиолетовая", color="blue", marker="s", linestyle="-.")

plt.title("Сравнение функций")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```
